## PUNTO 2 y 3

```python
P = np.array([[0.6,0.3,0.1],
              [0.2,0.6,0.2],
              [0.2,0.2,0.6]])
```

```python
print("Normal 0, Congestionada 1, Critica 2")
print(P)
```

```python
# Punto 2 - Literal C
P2_Normal_a_Critico = P[0][0]*P[0][2] + P[0][1]*P[1][2] + P[0][2]*P[2][2]
print(P2_Normal_a_Critico)
```

```python
# Punto 2 - Literal D
alpha_0 = np.array([[0.5,0.3,0.2]])

P2 = np.linalg.matrix_power(P,2)

#Probabilidades de estado
alpha_2 = alpha_0 @P2

print(alpha_2)
```

```python
# Punto 3 - Literal A
A = P.T - np.eye(3)
b = np.array([[0],[0],[1]])

A[-1, :] = 1

pi = np.linalg.solve(A, b)

print("Pi:")
print(pi)
print("Condición: El vector pi debe dar uno")
print(pi.sum())
```

```python
pi_2 = [0.33333333, 0.38888889, 0.27777778]

print(pi_2 @ P)
```

```python
# Punto 3 - Literal C
costo_esperado = 0.33333333*0 + 0.38888889*40000 + 0.27777778*160000
print(costo_esperado)
```

```python
# Punto 3 - Literal D

# Ya que es una cadena reducible y positiva recurrente, sabemos que
# miu (Tiempo medio de retorno) = 1 / p_i

# Para el caso de estado crítico
miu = 1/pi_2[2]

print(miu)
```

## PUNTO 5

```python
# Punto 5 - Literal B

p00 = 0.4**2 #Las dos siguen sin arreglarse
p01 = 2*0.6*0.4 #Una de las dos se arregla y la otra no
p02 = 0.6**2 #Se arreglan las dos
p10 = 0.2*0.4 #La que estaba buena falla y la mala sigue sin arreglarse
p11 = 0.8*0.4 + 0.2*0.6 #Ambas siguen como estaban o la mala se arregla y la buena falla
p12 = 0.8*0.6 #Sigue funcionando y la mala se arregla
p20 = 0.2**2 #Ambas
p21 = 2*(0.8*0.2) #Una de las dos sigue funcionando y la otra falla
p22 = 0.8**2 #Ambas siguen funcionando

print(p00,p01,p02)
print(p00+p01+p02)
print(p10,p11,p12)
print(p10+p11+p12)
print(p20,p21,p22)
print(p20+p21+p22)
```

```python
# Punto 5 - Literal C
P = np.array([[p00, p01, p02],
             [p10,p11,p12],
             [p20,p21,p22]])

print(P)
```
